# 1. search()

- Searches for the pattern anywhere in the string
- Returns only the first occurrence

- Syntax :
    - `re.search(pattern, string)`

- Return Type
  
  - If a match is found → Match Object

  - If no match is found → None

- 📌 Unlike findall(), it does not return a list

```python
import re

text = "Python is easy to learn"
result = re.search("easy", text)
print(result)
```

# 2. match()

- Matches the pattern only at the beginning of the string.
- Syntax :
    - `re.match(pattern, string)`

- Return Type
  
  - Match found → Match object

  - No match → None



```python
import re

re.match("Hello", "Hello World HEllo")   # Match
re.match("World", "Hello World")   # None
```

##3. findall()

- findall() is a function in Python’s re module used to search the entire string and extract all occurrences that match a given pattern
- Syntax :
    - `re.findall(pattern, string)`

- Return Type
    - Always returns a list
    - Each item in the list is a matched substring (string type)

<br>

### How does findall() work internally? (Deep Concept)

- When you run findall():

- Regex engine scans the entire string from left to right

- Every time a match is found → it is added to a result list

- The scan continues even after finding a match

- After finishing scanning the entire string, the list of matches is returned

```python
import re

s="Python is an oops lang.Python is also Fun prog lang python"

sub="Python"

words=re.findall(sub,s)

print(words)
```

# 4. finditer()

- finditer() is a function in Python’s re module

- It searches the entire string for a given pattern

- It returns all matches, but as an iterator of Match Objects

- Each match contains matched text + position information


- Syntax :
    - `re.finditer(pattern, string)`

- Return Type
  - Returns an iterator (iterable object)

  - Each element is a Match Object

  - If no match is found → returns an empty iterator

- 📌 Unlike findall(), it does not return a list

```python
import re

text = "Marks: 45, 67 and 89"

result = re.finditer(r"\d+", text)

for match in result:
    print(match)
```

# 5. sub()

- sub() is a function in Python’s re module

- It is used to replace occurrences of a pattern with another string

- It returns a new modified string

- The original string remains unchanged (strings are immutable)

- Syntax :

   - `re.finditer(pattern, string)`

- Return Type
  - Always returns a string

  - If no match is found → returns the original string unchanged

```python
import re

text = "I love Java"
result = re.sub("Java", "Python", text)
print(result)
```

# 6. split()

- split() is a function in Python’s re module

- It is used to split a string based on a regex pattern

- The pattern works as a delimiter

- It returns a list of substrings

- Syntax :     
  - `re.split(pattern, string)`

- Return Type :
  - Always returns a list

  - If the pattern is not found → returns a list containing the original string

```python
import re

text = "apple,banana,orange"
result = re.split(",", text)
print(result)
```

# Quantifiers in Regular Expressions
- Quantifiers in Regular Expressions are used to specify how many times a character, group, or pattern should occur.

- They help in searching number of occurrences of:

    - alphabets

    - digits

    - special symbols

- Quantifiers make regex flexible and powerful by controlling repetition.
- Quantifiers can be used with all regular method.

<br>

| Quantifier | Meaning         |
| ---------- | --------------- |
| `*`        | 0 or more       |
| `+`        | 1 or more       |
| `?`        | 0 or 1          |
| `{n}`      | exactly n       |
| `{n,}`     | at least n      |
| `{n,m}`    | between n and m |

<br>


<br><br>

# 1️⃣ Exactly One (k)

- Meaning

    - Matches exactly one occurrence of the character



```python
import re

re.findall("a", "banana")
```

# 2️⃣ One or More (+)
- Meaning

    - Matches one or more occurrences

    - At least one match is mandatory

```python
re.findall("a+", "acaaat")
```

# 3️⃣ Zero or More (*)
- Meaning

    - Matches zero or more occurrences

    - Match may or may not exist

```python
re.findall("a*", "baac")
```

# 4️⃣ Zero or One (?)
- Meaning

    - Matches zero or one occurrence

    - Used for optional characters

```python
re.findall("colou?r", "color colour")


# u is optional here.
```

# 5️⃣ Any Character (.)
- Meaning

    - Matches any single character

    - Except newline (\n) by default

```python
re.findall("c.t", "cat cot cut")
```

# 6️⃣ Curly Braces Quantifiers { } ⭐


## {n} – Exactly n times

```python
re.findall(r"\d{3}", "12345")
```

## {n,} – At least n times

```python
import re

re.findall(r"\d{2,}", "1 12 123 1234 as")
```

## {n,m} – Between n and m times

```python
re.findall(r"\d{2,3}", "1 12 123 1234")
```

## Greedy Nature of Quantifiers ⭐

- Quantifiers are greedy by default

- They match the maximum possible characters

```python
re.findall("a+", "aaaa")
```

## Lazy (Non-Greedy) Quantifiers

- Add ? after quantifier to make it lazy

- Matches the minimum possible characters

```python
re.findall("a+?", "aaaa")
```
